# MNIST-1D

```python
from mnist1d.data import make_dataset, get_dataset_args, get_templates
from mnist1d.utils import from_pickle, to_pickle, ObjectView, set_seed, plot_signals
###TODO padding###
arg_dict = ObjectView({
    'num_samples': 5000,
    'train_split': 0.8,
    'template_len': 12,
    'padding': [36,60],
    'scale_coeff': .4, 
    'max_translation': 48,
    'corr_noise_scale': 0.25,
    'iid_noise_scale': 2e-2,
    'shear_scale': 0.75,
    'shuffle_seq': False,
    'final_seq_length': 144,
    'seed': 42,
    'url': 'https://github.com/greydanus/mnist1d/raw/master/mnist1d_data.pkl'
})
data = make_dataset(arg_dict)

templates = get_templates()
print("Templates for the MNIST-1D dataset:")
x = templates['x']
t = templates['t']
y = templates['y']
fig = plot_signals(x, t, labels=y, args=arg_dict, ratio=1.33, dark_mode=True, do_transform=True)
```

```python
print(data.keys())
x, y, t = torch.tensor(data['x'],dtype=torch.float32).unsqueeze(-1), torch.tensor(data['y']), torch.tensor(data['t'])
x.shape, y.shape, t.shape
```

```python
#### With P learning ###
dt = 0.5 #ms
n_steps = arg_dict.final_seq_length

# init both teacher and student net
n_in, n_hidden1, n_hidden2, n_hidden3, n_hidden_ins, n_out = 1, 30, 30, 30, 60, 10
batch = 20
activation='tanh'
base_lr = 1e-2
Tau1 = dt+np.logspace(0, 2.1, n_hidden1, dtype=np.float32)
print(Tau1)
Tau2 = torch.tensor([1.2, 5., 10.]*10)
Tau3 = torch.tensor([1.3, 5.1, 11.]*10)
Tau4 = torch.tensor([1.4, 5.2, 12.]*10)

student = DENetwork(
    layers=torch.nn.ModuleList([
        FwdDENeurons(n_in=n_in, n_neurons=n_hidden1, tau = Tau1,
                     activation=activation, batch=batch, dt=dt, scale=1.6),
        FwdDENeurons(n_in=n_hidden1, n_neurons=n_hidden2, tau = Tau2,
                     activation='linear', batch=batch, dt=dt, scale=0.1),
        FwdDENeurons(n_in=n_hidden2, n_neurons=n_hidden2, tau = Tau3,
                     activation='linear', batch=batch, dt=dt, scale=0.1),
        LastFwdDENeurons(n_in=n_hidden2, n_neurons=n_hidden3, tau = Tau4,
                         activation=activation, batch=batch, dt=dt, scale=1.6),
        FwdNeurons(n_in=n_hidden3, n_neurons=n_hidden_ins, tau = dt,
                      activation=activation, batch=batch, dt=dt, scale=1.6),
        # FwdNeurons(n_in=n_hidden_ins, n_neurons=n_hidden_ins, tau = dt,
        #               activation=activation, batch=batch, dt=dt, scale=1.6),
        FwdNeurons(n_in=n_hidden_ins, n_neurons=n_out, tau = dt, 
                      activation='linear', batch=batch, dt=dt, beta=1.)
    ]), 
    batch=batch,
)

optimizer = torch.optim.Adam(
    student.parameters(),
    lr=base_lr
)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='min',
    factor=0.1,
    patience=10
)
```

```python
error_record, mismatch_record = [], []
Cum_errors = 0.3
num_epochs = 30

with torch.no_grad():
    for epoch in range(num_epochs):
        permu = np.random.permutation(4000).reshape(-1, batch)
        for i, idx in enumerate(permu):
            total_error = 0.
            one_hot_label = torch.zeros(batch, 10)   
            one_hot_label[np.arange(batch), y[idx]] = 1.
            sample = x[idx]
            student.reset()
            for t in range(n_steps):
                r_out, u_out = student.step(sample[:, t])
                if t>=136:
                    p = torch.softmax(r_out, dim=1)
                    error = one_hot_label - p
                    student.prop(error)
                    student.backwards()
                    optimizer.step()
                    optimizer.zero_grad()
                    total_error += -(one_hot_label * torch.log(p)).mean().item()
                else:
                    student.prop(0)
            
            Cum_errors = 0.98*Cum_errors + 0.02*total_error/8
            if (i*batch)%800==0:
                scheduler.step(Cum_errors)
                error_record.append(Cum_errors)   
                print(" error: %5e, iteration: %d samples"%(Cum_errors, i*batch))
                
print('finished')
```

```python
#TEST
x_test, y_test = torch.tensor(data['x_test'],dtype=torch.float32).unsqueeze(-1), torch.tensor(data['y_test'])
prediction = []
acc = 0.
with torch.no_grad():
    permu = np.random.permutation(1000).reshape(-1, batch)
    for i, idx in enumerate(permu):
        sample = x[idx]
        pred = torch.zeros(batch, 10)
        for t in range(n_steps):
            r_out,_= student.step(sample[:, t])
            if t>130:
                pred += r_out
        student.reset()
        prediction=torch.argmax(pred, dim=1)
        acc += ((prediction.numpy()==y[idx])*1.).sum()
acc /= 1000
print('Accuracy: ', acc)
```
