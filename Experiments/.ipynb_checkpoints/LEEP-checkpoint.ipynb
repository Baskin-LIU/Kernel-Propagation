### network

```python
class SequentialNetwork(torch.nn.Module):

    def __init__(self, layers, device="cpu"):
        super().__init__()
        self.layers = layers
        self.n_layer = len(layers)
        for i, l in enumerate(layers):
            if i != 0:
                l.set_previous_layer(layers[i-1])
            if i != self.n_layer-1:
                l.set_next_layer(layers[i+1])
            

    def step(self, r_in, noise=0, target=None):
        r = r_in
        for i, l in enumerate(self.layers):
            r, u = l.step(r, noise)

        return r, u

    def mismatch(self, layer=None):
        if layer is not None:
            return self.layers[layer].mismatch.clone().detach()
        else:
            return [l.mismatch for l in self.layers]

    def epsilon(self, layer=None):
        if layer is not None:
            return self.layers[layer].epsilon.clone().detach()
        else:
            return [l.epsilon for l in self.layers]

    def learnP(self, e_trg=0., learning=True):
        self.layers[-1].learnP(e_trg=e_trg)
        for l in reversed(self.layers[:-1]):
            l.learnP(e_trg=0)

    def learnW(self, layers=None):
        for i, l in enumerate(self.layers):
            l.learnW()

    def reset_P(self, ):
        for i, l in enumerate(self.layers):
            l.reset_P()

    def reset(self, ):
        for i, l in enumerate(self.layers):
            l.reset()

    def align(self, target):
        for i, l in enumerate(self.layers):
            l.u_bar = target.layers[i].u_bar.detach().clone()

    def align0(self,):
        for i, l in enumerate(self.layers):
            l.u_bar *= 0.
```

# Steady Input/$e^{trg}$ in Recurrent (rate) 

```python
dt = 0.5 # ms
init_t=200 # ms
test_t=0 # ms
duration = init_t+test_t
num_epoch = 1

# init both teacher and student net
n_class = 2
n = 4
n_in = n**2
n_hidden = 25
n_out = n_class
    
student = SequentialNetwork(
    layers=[
        RecNeurons(n_in=n_in, n_neurons=n_hidden, n_a=n_hidden, tau_d = (40, 60), inh=0.,
                   lr_p=3e0, lr_w = 1e-2, activation='sigmoid', dt=dt, beta=0.1),
        FwdNeurons(n_in=n_hidden, n_neurons=n_out, n_a=n_hidden, tau_d = dt, 
                   lr_p=3e0, lr_w = 1e-2, activation='linear', dt=dt, beta=0.1)
    ],)

#define input
input_fuc = gratingR(n, dt=dt)

#recorders
r_record, student_plot, rhod_record = [], [], []

target = 0
x = input_fuc(init_t+test_t, direction=target)
with torch.no_grad():
    for i in range(num_epoch):
        one_hot = F.one_hot(torch.tensor([[target]]), num_classes=n_class).float()
        student.reset()
        for step in range(int((init_t+test_t)//dt)): 
            r_out, u_out = student.step(x, noise=0.)
        
            
            #epsilon_plot.append(student.dismatch(0))
            rhod_record.append(student.layers[0].rho.d)
            student_plot.append(u_out)
            r_record.append(student.layers[0].r)
            e_trg = (one_hot - torch.sigmoid(u_out+70))[0, 0] if step*dt>init_t else 0
            student.learnP(e_trg=e_trg)

        print('epoch %d done'%i)

r_record = torch.vstack(r_record)
rhod_record = torch.vstack(rhod_record)
```

# Steady Input/$e^{trg}$ in Recurrent (spike) 

```python
dt = 0.5 # ms
init_t=120 # ms
test_t=0 # ms
duration = init_t+test_t
num_epoch = 1

# init both teacher and student net
n_class = 2
n = 4
n_in = n**2
n_hidden = 25
n_out = n_class
    
student = SequentialNetwork(
    layers=[
        RecSpkNeurons(n_in=n_in, n_neurons=n_hidden, n_a=n_hidden, tau_d = (40, 60), inh=4., threshold=0.1,
                   lr_p=3e0, lr_w = 1e-2, activation='spiking', dt=dt, beta=0.1),
        FwdSpkNeurons(n_in=n_hidden, n_neurons=n_out, n_a=n_hidden, tau_d = dt, 
                   lr_p=3e0, lr_w = 1e-2, activation='linear', dt=dt, beta=0.1)
    ],)

#define input
input_fuc = gratingR(n, dt=dt)

#recorders
u_record, r_record, student_plot, rhod_record = [], [], [], []

target = 0
x = 0.15*input_fuc(init_t+test_t, direction=target)
with torch.no_grad():
    for i in range(num_epoch):
        one_hot = F.one_hot(torch.tensor([[target]]), num_classes=n_class).float()
        for step in range(int((init_t+test_t)//dt)): 
            r_out, u_out = student.step(x, noise=0.)
    
            #epsilon_plot.append(student.dismatch(0))
            rhod_record.append(student.layers[0].rho.d)
            student_plot.append(u_out)
            r_record.append(student.layers[0].r)
            u_record.append(student.layers[0].u_bar.clone().detach())
            e_trg = (one_hot - torch.sigmoid(u_out+70))[0, 0] if step*dt>init_t else 0
            student.learnP(e_trg=e_trg)
        #student.reset()
        print('epoch %d done'%i)

r_record = torch.vstack(r_record)
rhod_record = torch.vstack(rhod_record)
u_record = torch.vstack(u_record)
```

```python
spike_matrix = r_record.T
steps=spike_matrix.shape[1]
n_neurons = spike_matrix.shape[0]
time = np.linspace(0, steps*dt, steps)

fig, ax = plt.subplots(figsize=(14,10))

# Extract spike times
for neuron_id in range(n_neurons):
    spike_times = time[spike_matrix[neuron_id]>0]
    ax.scatter(spike_times, [neuron_id]*len(spike_times), 
               s=10, color="black")

ax.set_xlabel("Time (s)")
ax.set_ylabel("Neuron")
ax.set_title("Spike Raster Plot")
ax.set_ylim(-0.5, n_neurons-0.5)  # keep y axis clean
ax.set_yticks(np.arange(n_neurons))
#ax2 = ax.twinx()
#ax2.plot(vol_record)

plt.show()
```

```python
wTe_trg = 0.1*student.layers[1].W_in.T @ torch.tensor([-0.5, 0.5])
dt_tau = student.layers[0].dt_tau.numpy()
W_r = student.layers[0].W_r
mismatch_record = [torch.zeros(n_hidden),]
epsilon_true = []
for t in reversed(range(rhod_record.shape[0])):
    if t>80:
        epsilon_true.insert(0, rhod_record[t]*(wTe_trg + W_r@mismatch_record[0]))
    else:
        epsilon_true.insert(0, rhod_record[t]*(W_r@mismatch_record[0]))
    mismatch = torch.zeros(n_hidden)
    for i, epsi in enumerate(epsilon_true):
        mismatch += epsi*dt_tau* (1-dt_tau)**i
    mismatch_record.insert(0, mismatch)
mismatch_record = torch.vstack(mismatch_record[:-1])
```

## Hidden layer 

```python
#### With P learning ###
dt = 0.1 #ms

init_phase = 200 #ms
init_phase_steps = int(init_phase/dt) 
learn_phase = 8000 # ms
learn_phase_steps = int(learn_phase/dt) 
eval_phase = 500 # ms
eval_phase_steps = int(eval_phase/dt)

# init both teacher and student net
n_in, n_hidden, n_out = 4, 2, 1   

student = SequentialNetwork(
    layers=[
        FwdNeurons(n_in=n_in, n_neurons=n_hidden, n_a=n_in, tau_d = [17, 67], lr_p=2e-3, lr_w = 9e-1,
                     activation='linear', afunc='inp_b', P_bias=True, dt=dt, beta=0.1),
        FwdNeurons(n_in=n_hidden, n_neurons=n_out, n_a=n_hidden, tau_d = dt, 
                   lr_p=1e-1, lr_w = 4e-1, activation='linear', afunc='inp_b', P_bias=True, dt=dt, beta=0.1)
    ],)



teacher = copy.deepcopy(student)
student_naive = copy.deepcopy(student)

# deviate teacher by small perturbation
for i, l in enumerate(student.layers[:-1]):
    deviate = torch.randn(l.W_in.shape)*2e-1
    teacher.layers[i].W_in += deviate

#define input
input_fuc = sin_function(n_in, period=(10, 80), bias = (0, 0.5), magnitude=(1, 2), phase = (0, 6.2), dt=dt)

#recorders
mismatch_record, error_record = [], []

Rcd = Recorder(['input', 'teacher', 'student', 'mismatch'])

with torch.no_grad():
    
    for t in range(init_phase_steps):
        input = input_fuc(t)
        r_out_target, u_out_target = teacher.step(input)
        r_out, u_out = student.step(input)
        Rcd.rec([input, u_out_target, u_out, student.mismatch(0)])
        
    error = u_out_target - u_out

    print("reach steady state with error: %f"%(error**2).mean().item())
    
Rcd.finish()

fig, axs = plt.subplots(2, 1, figsize=(10, 8), sharex=True)
x = np.arange(0, init_phase, dt)
axs[0].plot(x, Rcd['input'].mean(axis=0), color='r', linewidth=2)
for i in range(n_in):
    axs[0].plot(x, Rcd['input'][i][0:], alpha = 0.3)
for i in range(n_out):
    axs[1].plot(x, Rcd['teacher'][i][0:], label="teacher", color = black, linewidth=2)
    axs[1].plot(x, Rcd['student'][i][0:], label="student", color = lightblue, linewidth=2)
axs[1].legend()
axs[1].set_xlabel("t (ms)")

plt.show()
```

```python
cum_error = (error**2).mean().item()
cum_eps = 0
Rcd = Recorder(['mismatch', 'epsilon'])
with torch.no_grad():
    for t in range(learn_phase_steps):
        t += init_phase_steps
        input = input_fuc(t)
        r_out_target, u_out_target = teacher.step(input)
        r_out, u_out = student.step(input)
        error = u_out_target - u_out
        student.learnP(error)

        Rcd.rec([student.mismatch(0), student.epsilon(0)])

        cum_error = 0.999*cum_error + 0.001*(error**2).mean().item()
        cum_eps = 0.99*cum_eps + 0.01*(student.mismatch(0)**2).mean().item()
        
        if cum_error < 1e-5:
            print("finished with %d steps"%t)
            break
        if (t*dt)%100==0:
            mismatch_record.append(cum_eps)
            error_record.append(cum_error)
            
            print("target: ", u_out_target.mean().numpy(), "output: ", u_out.mean().numpy(), 
                  " epsilon: %5e, error: %5e, t: %d ms"%(mismatch_record[-1], error_record[-1], t*dt))

Rcd.finish()
```

### Check P learning
Compare e with true e

```python
epsilon_tilde = np.zeros(Rcd['epsilon'].shape)
dt_tau = student.layers[0].dt_tau.unsqueeze(1).numpy()
for i in range(int(1000/dt)):
    epsilon_tilde[:,:-i-1] += Rcd['epsilon'][:, i+1:] * (1-dt_tau)**i
epsilon_tilde *= dt_tau

start=60000
end = start+2000
time=np.arange(start, end)*dt

fig, axs = plt.subplots(2, 1, sharex=True, figsize=(10, 8))
for ind in range(n_hidden):
    axs[ind].plot(time, epsilon_tilde[ind, start:end], label=r"$\mathbf{e}$", linewidth=3.)
    axs[ind].plot(time, Rcd['mismatch'][ind, start:end], label=r"$\mathbf{u}^{ɑ}$", linewidth=3.)
    axs[ind].set_title("hidden neuron %d"%ind)
    axs[ind].spines['top'].set_visible(False)
    axs[ind].spines['right'].set_visible(False)
    

axs[1].set_xlabel('t(ms)')
axs[1].legend(loc='upper right', frameon=False)

plt.tight_layout()
plt.savefig('./fig/Plearnhid.png')
plt.show()
```

```python
#### With P learning ###
dt = 0.1 #ms

init_phase = 200 #ms
init_phase_steps = int(init_phase/dt) 
learn_phase = 8000 # ms
learn_phase_steps = int(learn_phase/dt) 
eval_phase = 500 # ms
eval_phase_steps = int(eval_phase/dt)

# init both teacher and student net
n_in, n_hidden, n_out = 4, 2, 1
beta=0.1
activation='sigmoid'

student = SequentialNetwork(
    layers=[
        FwdNeurons(n_in=n_in, n_neurons=n_hidden, n_a=n_in, tau_d = (20, 40), lr_p=5e-3, lr_w = 1e-3,
                     activation=activation, afunc='inp_b', P_bias=True, dt=dt, beta=beta),
        FwdNeurons(n_in=n_hidden, n_neurons=n_hidden, n_a=n_hidden, tau_d = (20, 40), lr_p=5e-3, lr_w = 1e-3,
                     activation=activation, afunc='inp_b', P_bias=True, dt=dt, beta=beta),
        FwdNeurons(n_in=n_hidden, n_neurons=n_out, n_a=n_hidden, tau_d = dt, 
                   lr_p=2e-1, lr_w = 1e-5, activation='linear', afunc='inp_b', P_bias=True, dt=dt, beta=beta)
    ],)


teacher = copy.deepcopy(student)
student_naive = copy.deepcopy(student)

# deviate teacher by small perturbation
for i, l in enumerate(student.layers[:-1]):
    deviate = torch.randn(l.W_in.shape)*2e-1
    teacher.layers[i].W_in += deviate

#define input
input_fuc = mul_sin_function(n=n_in, n_sin=2, period=(20, 80), bias = (-0.5, 0.5), magnitude=(1, 2), phase = (0, 6), dt=0.1)

#recorders
mismatch_record, error_record = [], []
Rcd = Recorder(['input', 'teacher', 'student', 'mismatch'])

with torch.no_grad():
    
    for t in range(init_phase_steps):
        input = input_fuc(t)
        r_out_target, u_out_target = teacher.step(input)
        r_out, u_out = student.step(input)
        Rcd.rec([input, u_out_target, u_out, student.mismatch(0)])
        
    error = u_out_target - u_out
    
    print("reach steady state with error: %f"%(error**2).mean().item())

Rcd.finish()

fig, axs = plt.subplots(2, 1, figsize=(10, 8), sharex=True)
x = np.arange(0, init_phase, dt)
axs[0].plot(x, Rcd['input'].mean(axis=0), color='r', linewidth=2)
for i in range(n_in):
    axs[0].plot(x, Rcd['input'][i][0:], alpha = 0.3)
for i in range(n_out):
    axs[1].plot(x, Rcd['teacher'][i][0:], label="teacher", color = black, linewidth=2.)
    axs[1].plot(x, Rcd['student'][i][0:], label="student", color = lightblue, linewidth=2.)
axs[1].legend()
axs[1].set_xlabel("t (ms)")
axs[0].set_ylabel(r"$r_{in}$")
axs[1].set_ylabel(r"$r_o$")

for ax in axs:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    
plt.tight_layout()
#plt.savefig('./fig/TechStuInit%s.png'%activation)
plt.show()
```

```python
cum_error = (error**2).mean().item()
cum_eps = 0
Rcd = Recorder(['input', 'r1', 'r2','mismatch', 'mismatch2', 'epsilon2', 'rhod'])
with torch.no_grad():
    for t in range(learn_phase_steps):
        t += init_phase_steps
        input = input_fuc(t)
        r_out_target, u_out_target = teacher.step(input)
        r_out, u_out = student.step(input)
        error = u_out_target - u_out
        student.learnP(error)
        #student.learnW()

        Rcd.rec([input, student.layers[0].r, student.layers[1].r, student.mismatch(0),
                 student.mismatch(1), student.epsilon(1), student.layers[0].rho.d])

        cum_error = 0.999*cum_error + 0.001*(error**2).mean().item()
        cum_eps = 0.99*cum_eps + 0.01*(student.mismatch(0).abs()).mean().item()
        
        if (t*dt)%100==0:
            mismatch_record.append(cum_eps)
            error_record.append(cum_error)
            
            print("target: ", u_out_target.mean().numpy(), "output: ", u_out.mean().numpy(), 
                  " mismatch: %5e, error: %5e, t: %d ms"%(mismatch_record[-1], error_record[-1], t*dt))

Rcd.finish()
```

```python
#### With P learning ###
dt = 0.1 #ms

init_phase = 200 #ms
init_phase_steps = int(init_phase/dt) 
learn_phase = 18000 # ms
learn_phase_steps = int(learn_phase/dt) 
eval_phase = 200 # ms
eval_phase_steps = int(eval_phase/dt)

# init both teacher and student net
n_in, n_hidden1, n_hidden2, n_out = 4, 3, 2, 1
beta=0.05
activation='sigmoid'

# student = SequentialNetwork(
#     layers=[
#         FwdNeurons(n_in=n_in, n_neurons=n_hidden1, n_a=n_in, tau_d = [dt, 1, 2], lr_p=4e-2, lr_w = 0,
#                      activation=activation, afunc='all', P_bias=True, dt=dt),
#         FwdNeurons(n_in=n_hidden1, n_neurons=n_hidden2, n_a=n_hidden1, tau_d = [13, 67], lr_p=4e-2, lr_w = 2.0,
#                      activation=activation, afunc='all', P_bias=True, dt=dt),
#         FwdNeurons(n_in=n_hidden2, n_neurons=n_out, n_a=n_hidden2, tau_d = dt, 
#                    lr_p=2e-1, lr_w = 0, activation='linear', afunc='inp_b', P_bias=True, dt=dt, beta=beta)
#     ],)

student = SequentialNetwork(
    layers=[
        FwdEPNeurons(n_in=n_in, n_neurons=n_hidden1, tau_d = [dt, 2., 3.3], lr_w = 0.8,
                     activation=activation, dt=dt),
        FwdEPNeurons(n_in=n_hidden1, n_neurons=n_hidden2, tau_d = [7, 13], lr_w = 0.8,
                     activation=activation, dt=dt),
        FwdLENeurons(n_in=n_hidden2, n_neurons=n_out, tau_d = dt, 
                   lr_w=0, activation='linear', dt=dt, beta=beta)
    ],)
student_naive = copy.deepcopy(student)

student2 = SequentialNetwork(
    layers=[
        FwdLENeurons(n_in=n_in, n_neurons=n_hidden1, tau_d = student.layers[0].tau_d, lr_w = 0.8,
                      W_in=student.layers[0].W_in.detach().clone(), activation=activation, dt=dt),
        FwdLENeurons(n_in=n_hidden1, n_neurons=n_hidden2, tau_d=student.layers[1].tau_d, lr_w = 0.8,
                      W_in=student.layers[1].W_in.detach().clone(), activation=activation, dt=dt),
        FwdLENeurons(n_in=n_hidden2, n_neurons=n_out, tau_d=dt, lr_w=0,
                     W_in=student.layers[-1].W_in.detach().clone(), activation='linear', dt=dt, beta=beta)
    ],)

student3 = SequentialNetwork(
    layers=[
        FwdGLENeurons(n_in=n_in, n_neurons=n_hidden1, tau_d = student.layers[0].tau_d, lr_w = 0.8,
                      W_in =student.layers[0].W_in.detach().clone(), activation=activation, dt=dt),
        FwdGLENeurons(n_in=n_hidden1, n_neurons=n_hidden2, tau_d = student.layers[1].tau_d, lr_w = 0.8,
                      W_in =student.layers[1].W_in.detach().clone(), activation=activation, dt=dt),
        FwdGLENeurons(n_in=n_hidden2, n_neurons=n_out, tau_d=dt, lr_w = 0,
                      W_in=student.layers[-1].W_in.detach().clone(), activation='linear', dt=dt, beta=beta)
    ],)

Students={'tNLA':student, 'LE':student2, 'GLE':student3, 'naive':student_naive}

teacher = copy.deepcopy(student)

# deviate teacher by small perturbation
for i, l in enumerate(student.layers[:-1]):
    deviate = torch.randn(l.W_in.shape)*6e-1
    teacher.layers[i].W_in += deviate

# teacher = SequentialNetwork(
#     layers=[
#         FwdLENeurons(n_in=n_in, n_neurons=n_hidden1, tau_d = student.layers[0].tau_d, lr_w = 0.,
#                       W_in=student.layers[0].W_in.detach().clone(), activation=activation, dt=dt),
#         FwdLENeurons(n_in=n_hidden1, n_neurons=n_hidden2, tau_d=student.layers[1].tau_d, lr_w = 2.0,
#                       W_in=student.layers[1].W_in.detach().clone(), activation=activation, dt=dt),
#         FwdLENeurons(n_in=n_hidden2, n_neurons=n_out, tau_d=dt, lr_w=0,
#                      W_in=student.layers[-1].W_in.detach().clone(), activation='linear', dt=dt, beta=beta)
#     ],)



input_fuc = mul_sin_function(n=n_in, n_sin=2, period=(5, 10), bias = (-0.1, 0.1), magnitude=(0.2, 1.0), phase = (0, 6), dt=0.1)

#recorders
Rcd = Recorder(['input', 'teacher', 'student', 'mismatch'])

with torch.no_grad():
    
    for t in range(init_phase_steps):
        input = input_fuc(t)
        r_out_target, u_out_target = teacher.step(input)
        r_out, u_out = student.step(input)
        Rcd.rec([input, u_out_target, u_out, student.mismatch(0)])
        
    error = u_out_target - u_out
    print("reach steady state with error: %f"%(error**2).mean().item())

Rcd.finish()

fig, axs = plt.subplots(2, 1, figsize=(10, 8), sharex=True)
x = np.arange(0, init_phase, dt)
axs[0].plot(x, Rcd['input'].mean(axis=0), color='r', linewidth=2)
for i in range(n_in):
    axs[0].plot(x, Rcd['input'][i][0:], alpha = 0.3)
for i in range(n_out):
    axs[1].plot(x, Rcd['teacher'][i][0:], label="teacher", color = black)
    axs[1].plot(x, Rcd['student'][i][0:], label="student", color = lightblue)
axs[1].legend()
axs[1].set_xlabel("t (ms)")
axs[0].set_ylabel(r"$r_{in}$")
axs[1].set_ylabel(r"$r_o$")

plt.show()
```

```python
Cum_errors = 3*[(error**2).mean().item()]
error_record, mismatch_record = [], []
cum_eps = 0
Students['LE'].align(student)
Students['GLE'].align(student)
Rcd = Recorder(['input', 'mismatch', 'mismatch2', 'epsilon2', 'rhod', 'epsilon_breve'])

with torch.no_grad():
    for t in range(learn_phase_steps):
        t += init_phase_steps
        input = input_fuc(t)
        r_out_target, u_out_target = teacher.step(input)
        
        for i, stu in enumerate(['tNLA', 'LE', 'GLE']):
            r_out, u_out = Students[stu].step(input)
            error = u_out_target - u_out
            Students[stu].learnP(error)
            Students[stu].learnW()
            Cum_errors[i] = 0.999*Cum_errors[i] + 0.001*(error**2).mean().item()
        # Rcd.rec([input, student.mismatch(0), student.mismatch(-1), student.epsilon(1), 
        #          student.layers[0].rho.d, student3.mismatch(0)])
        cum_eps = 0.999*cum_eps + 0.001*(Students['tNLA'].mismatch(0).abs()).mean().item()
        
        if (t*dt)%100==0:
            mismatch_record.append(cum_eps)
            error_record.append(list(Cum_errors))
            
            print("target: ", u_out_target.mean().numpy(), "output: ", u_out.mean().numpy(), 
                  " mismatch: %5e, error: %5e, t: %d ms"%(mismatch_record[-1], Cum_errors[0], t*dt))

#Rcd.finish()
error_record = np.array(error_record).T
```

## Single Recurrent layer in Parity

```python
#### With P learning ###
dt = 0.1 #ms

init_phase = 100 #ms
init_phase_steps = int(init_phase/dt) 
learn_phase = 7000 # ms
learn_phase_steps = int(learn_phase/dt) 
eval_phase = 100 # ms
eval_phase_steps = int(eval_phase/dt)

# init both teacher and student net
n_in, n_rec, n_hidden, n_out = 1, 20, 8, 2
beta=4e-1
activation ='sigmoid'
N, bit_time = 3, 4

lr_in, lr_o = 1., 0.5

student = SequentialNetwork(
    layers=[
        RecLEEPNeurons(n_in=n_in, n_neurons=n_rec, tau_d = dt*np.logspace(0, 3.2, n_rec, dtype=np.float32), lr_w=lr_in, inh=0.4,
                     activation=activation, dt=dt, beta=beta),
        FwdLENeurons(n_in=n_rec, n_neurons=n_hidden, tau_d=dt, 
                    lr_w=lr_o, activation='sigmoid', dt=dt, beta=beta),
        FwdLENeurons(n_in=n_hidden, n_neurons=n_out, tau_d=dt, 
                    lr_w=lr_o, activation='linear', dt=dt, beta=beta)
    ],)

student_naive = copy.deepcopy(student)

student2 = SequentialNetwork(
    layers=[
        RecLEEP1Neurons(n_in=n_in, n_neurons=n_rec, tau_d=student_naive.layers[0].tau_d, lr_w=lr_in,
                     W_in=student_naive.layers[0].W_in.detach().clone(), W_r=student_naive.layers[0].W_r.detach().clone(),
                     activation=activation, dt=dt),
        FwdLENeurons(n_in=n_rec, n_neurons=n_hidden, tau_d=dt, W_in=student_naive.layers[1].W_in.detach().clone(),
                    lr_w=lr_o, activation='sigmoid', dt=dt, beta=beta),
        FwdLENeurons(n_in=n_hidden, n_neurons=n_out, tau_d=dt, lr_w=lr_o,
                     W_in=student_naive.layers[-1].W_in.detach().clone(), activation='linear', dt=dt, beta=beta)
    ],)

student3 = SequentialNetwork(
    layers=[
        RecLEEP2Neurons(n_in=n_in, n_neurons=n_rec, tau_d=student_naive.layers[0].tau_d, lr_w=lr_in,
                      W_in =student_naive.layers[0].W_in.detach().clone(), W_r=student_naive.layers[0].W_r.detach().clone(),
                      activation=activation, dt=dt),
        FwdLENeurons(n_in=n_rec, n_neurons=n_hidden, tau_d=dt, W_in=student_naive.layers[1].W_in.detach().clone(),
                    lr_w=lr_o, activation='sigmoid', dt=dt, beta=beta),
        FwdLENeurons(n_in=n_hidden, n_neurons=n_out, tau_d=dt, lr_w=lr_o,
                      W_in=student_naive.layers[-1].W_in.detach().clone(), activation='linear', dt=dt, beta=beta)
    ],)

student4 = SequentialNetwork(
    layers=[
        RecEPNeurons(n_in=n_in, n_neurons=n_rec, tau_d=student_naive.layers[0].tau_d, lr_w=lr_in,
                     W_in=student_naive.layers[0].W_in.clone(), W_r=student_naive.layers[0].W_r.clone(),
                     activation=activation, dt=dt),
        FwdLENeurons(n_in=n_rec, n_neurons=n_hidden, tau_d=dt, W_in=student_naive.layers[1].W_in.detach().clone(),
                    lr_w=lr_o, activation='sigmoid', dt=dt, beta=beta),
        FwdLENeurons(n_in=n_hidden, n_neurons=n_out, tau_d=dt, lr_w=lr_o,
                     W_in=student_naive.layers[-1].W_in.clone(), activation='linear', dt=dt, beta=beta)
    ],)

Students={'LEEP':student, 'LEEP1':student2, 'LEEP2':student3, 'eProp':student4, 'naive':student_naive}

input_fuc = StepParity(N=N, bit_time=bit_time, dt=dt)

#recorders
Rcd = Recorder(['input', 'teacher', 'student', 'mismatch'])

with torch.no_grad():
    
    for t in range(init_phase_steps):
        input, label = input_fuc()
        r_out, u_out = student.step(input)
        Rcd.rec([input, label, u_out, student.mismatch(0)])
        
    error = label - torch.sigmoid(u_out)
    print("reach steady state with error: %f"%(error**2).mean().item())

Rcd.finish()

fig, axs = plt.subplots(3, 1, figsize=(10, 10), sharex=True)
x = np.arange(0, init_phase, dt)
axs[0].plot(x, Rcd['input'].mean(axis=0)*0.6, color=black, label='input', linewidth=2, alpha=0.8)
axs[0].plot(x, 0.5*(Rcd['teacher'][1]-Rcd['teacher'][0]), label="Parity", color = red, alpha=0.8, linewidth=3)
for i in range(n_out):
    axs[i+1].plot(x, Rcd['student'][i], label="student %d"%i, color = lightblue)


axs[-1].set_xlabel("t (ms)")
axs[0].set_ylabel(r"$r_{in}$")
axs[0].legend()
axs[1].set_ylabel(r"$r_o$")

plt.show()
```

```python
Cum_errors = np.array(4*[(error**2).mean().item()])
error_record, mismatch_record = [], []
cum_eps = 0
Students['LEEP1'].align(student)
Students['LEEP2'].align(student)
Students['eProp'].align(student)
Rcd = Recorder(['input', 'mismatch', 'mismatch2', 'epsilon2', 'rhod', 'epsilon_breve'])

with torch.no_grad():
    for t in range(learn_phase_steps):
        t += init_phase_steps
        input, label = input_fuc()
        
        for i, stu in enumerate(['LEEP', 'LEEP1', 'LEEP2', 'eProp']):
            r_out, u_out = Students[stu].step(input)
            error = (label - torch.sigmoid(u_out))
            Students[stu].learnP(error)
            Students[stu].learnW()
            Cum_errors[i] = 0.999*Cum_errors[i] + 0.001*(error**2).mean().item()

        # Rcd.rec([input, student.mismatch(0), student.mismatch(-1), student.epsilon(1), 
        #          student.layers[0].rho.d, student3.mismatch(0)])
        cum_eps = 0.99*cum_eps + 0.01*(Students['LEEP'].mismatch(0).abs()).mean().item()

        if (t*dt)%100==0:
            mismatch_record.append(cum_eps)
            error_record.append(list(Cum_errors))
            
            for i, stu in enumerate(['LEEP', 'LEEP1', 'LEEP2', 'eProp']):
                Students[stu].layers[-1].beta *= 0.95
            
            print(" mismatch: %5e, error: %5e, t: %d ms"%(mismatch_record[-1], Cum_errors[0], t*dt))

Rcd.finish()
error_record = np.array(error_record).T
```

## Single Recurrent layer in N-Sum

```python
#### With P learning ###
dt = 0.1 #ms

init_phase = 100 #ms
init_phase_steps = int(init_phase/dt) 
learn_phase = 10000 # ms
learn_phase_steps = int(learn_phase/dt) 
eval_phase = 100 # ms
eval_phase_steps = int(eval_phase/dt)

# init both teacher and student net
n_in, n_rec, n_hidden, n_out = 1, 10, 6, 1
beta=2e-2
activation ='sigmoid'
N, bit_time = 3, 4
input_fuc = StepSum(N=N, bit_time=bit_time, dt=dt)

lr_in, lr_o = 1., 0.2

student = SequentialNetwork(
    layers=[
        RecLEEPNeurons(n_in=n_in, n_neurons=n_rec, tau_d = dt*np.logspace(0, 2.5, n_rec, dtype=np.float32), lr_w=lr_in, inh=0.4,
                     activation=activation, dt=dt, beta=beta),
        FwdLENeurons(n_in=n_rec, n_neurons=n_hidden, tau_d=dt, 
                    lr_w=lr_o, activation='sigmoid', dt=dt, beta=beta),
        FwdLENeurons(n_in=n_hidden, n_neurons=n_out, tau_d=dt, 
                    lr_w=lr_o, activation='linear', dt=dt, beta=beta)
    ],)

student_naive = copy.deepcopy(student)

student2 = SequentialNetwork(
    layers=[
        RecLEEP1Neurons(n_in=n_in, n_neurons=n_rec, tau_d=student_naive.layers[0].tau_d, lr_w=lr_in,
                     W_in=student_naive.layers[0].W_in.detach().clone(), W_r=student_naive.layers[0].W_r.detach().clone(),
                     activation=activation, dt=dt),
        FwdLENeurons(n_in=n_rec, n_neurons=n_hidden, tau_d=dt, W_in=student_naive.layers[1].W_in.detach().clone(),
                    lr_w=lr_o, activation='sigmoid', dt=dt, beta=beta),
        FwdLENeurons(n_in=n_hidden, n_neurons=n_out, tau_d=dt, lr_w=lr_o,
                     W_in=student_naive.layers[-1].W_in.detach().clone(), activation='linear', dt=dt, beta=beta)
    ],)

student3 = SequentialNetwork(
    layers=[
        RecLEEP2Neurons(n_in=n_in, n_neurons=n_rec, tau_d=student_naive.layers[0].tau_d, lr_w=lr_in,
                      W_in =student_naive.layers[0].W_in.detach().clone(), W_r=student_naive.layers[0].W_r.detach().clone(),
                      activation=activation, dt=dt),
        FwdLENeurons(n_in=n_rec, n_neurons=n_hidden, tau_d=dt, W_in=student_naive.layers[1].W_in.detach().clone(),
                    lr_w=lr_o, activation='sigmoid', dt=dt, beta=beta),
        FwdLENeurons(n_in=n_hidden, n_neurons=n_out, tau_d=dt, lr_w=lr_o,
                      W_in=student_naive.layers[-1].W_in.detach().clone(), activation='linear', dt=dt, beta=beta)
    ],)

student4 = SequentialNetwork(
    layers=[
        RecEPNeurons(n_in=n_in, n_neurons=n_rec, tau_d=student_naive.layers[0].tau_d, lr_w=lr_in,
                     W_in=student_naive.layers[0].W_in.clone(), W_r=student_naive.layers[0].W_r.clone(),
                     activation=activation, dt=dt),
        FwdLENeurons(n_in=n_rec, n_neurons=n_hidden, tau_d=dt, W_in=student_naive.layers[1].W_in.detach().clone(),
                    lr_w=lr_o, activation='sigmoid', dt=dt, beta=beta),
        FwdLENeurons(n_in=n_hidden, n_neurons=n_out, tau_d=dt, lr_w=lr_o,
                     W_in=student_naive.layers[-1].W_in.clone(), activation='linear', dt=dt, beta=beta)
    ],)

Students={'LEEP':student, 'LEEP1':student2, 'LEEP2':student3, 'eProp':student4, 'naive':student_naive}



#recorders
Rcd = Recorder(['input', 'teacher', 'student', 'mismatch'])

with torch.no_grad():
    
    for t in range(init_phase_steps):
        input, label = input_fuc()
        r_out, u_out = student.step(input)
        Rcd.rec([input, label, u_out, student.mismatch(0)])
        
    error = label - u_out
    print("reach steady state with error: %f"%(error**2).mean().item())

Rcd.finish()

fig, axs = plt.subplots(n_out+1, 1, figsize=(10, 10), sharex=True)
x = np.arange(0, init_phase, dt)
axs[0].plot(x, Rcd['input'].mean(axis=0)*0.6, color=black, label='input', linewidth=2, alpha=0.8)
axs[0].plot(x, Rcd['teacher'][0], label="Sum", color = red, alpha=0.8, linewidth=3)
for i in range(n_out):
    axs[i+1].plot(x, Rcd['student'][i], label="student %d"%i, color = lightblue)


axs[-1].set_xlabel("t (ms)")
axs[0].set_ylabel(r"$r_{in}$")
axs[0].legend()
axs[1].set_ylabel(r"$r_o$")

plt.show()
```

```python
Cum_errors = np.array(4*[(error**2).mean().item()])
error_record, mismatch_record = [], []
cum_eps = 0
Students['LEEP1'].align(student)
Students['LEEP2'].align(student)
Students['eProp'].align(student)
Rcd = Recorder(['input', 'mismatch', 'mismatch2', 'epsilon2', 'rhod', 'epsilon_breve'])

with torch.no_grad():
    for t in range(learn_phase_steps):
        t += init_phase_steps
        input, label = input_fuc()
        
        for i, stu in enumerate(['LEEP', 'LEEP1', 'LEEP2', 'eProp']):
            r_out, u_out = Students[stu].step(input)
            error = label - u_out
            Students[stu].learnP(error)
            Students[stu].learnW()
            Cum_errors[i] = 0.999*Cum_errors[i] + 0.001*(error**2).mean().item()

        # Rcd.rec([input, student.mismatch(0), student.mismatch(-1), student.epsilon(1), 
        #          student.layers[0].rho.d, student3.mismatch(0)])
        cum_eps = 0.99*cum_eps + 0.01*(Students['LEEP'].mismatch(0).abs()).mean().item()

        if (t*dt)%100==0:
            mismatch_record.append(cum_eps)
            error_record.append(list(Cum_errors))
            
            for i, stu in enumerate(['LEEP', 'LEEP1', 'LEEP2', 'eProp']):
                Students[stu].layers[-1].beta *= 0.95
            
            print(" mismatch: %5e, error: %5e, t: %d ms"%(mismatch_record[-1], Cum_errors[0], t*dt))

Rcd.finish()
error_record = np.array(error_record).T
```

```python
fig = plt.figure(figsize=(10, 4))
time = np.arange(error_record.shape[1])*100
ax = fig.add_subplot(111)
ax2 = ax.twinx()
ax.set_yscale('log')
ax.set_ylabel('loss', rotation=0)
ax2.set_ylabel(r"$|\mathbf{u}^{ɑ}|$", rotation=0)
ax2.set_yscale('log')
l1,  = ax.plot(time, error_record[0], label="LEEP", color=blue, alpha=0.8)
l11, = ax.plot(time, error_record[1], label="LEEP1", color=green, alpha=0.8)
l12, = ax.plot(time, error_record[2], label="LEEP2", color=purple, alpha=0.8)
l13, = ax.plot(time, error_record[3], label="eProp", color=pink, alpha=0.8)
l2, = ax2.plot(time, mismatch_record, label = r"$|\mathbf{u}^{ɑ}|$", color=brown, linestyle='--', linewidth=3., alpha=0.7)
lines = [l1, l11, l12, l13, l2]
labels = [line.get_label() for line in lines]
ax.legend(lines, labels, loc='upper right', frameon=False)
ax.set_xlabel('t (ms)')

plt.tight_layout()
#plt.savefig('./fig/losscurvecompareRec.png')
plt.show()
```

```python
Rcd_eval = Recorder(['input', 'teacher', 'LEEP', 'LEEP1', 'LEEP2', 'eProp', 'naive', 'LEEPr', 'LEEP1r', 'LEEP2r', 'ePropr','naiver'])

#Students['naive'].align(student)
with torch.no_grad():
    for t in range(eval_phase_steps):
        t = t+init_phase_steps+learn_phase_steps
        input, label = input_fuc()
        Rcd_eval.rec_single('input', input)
        Rcd_eval.rec_single('teacher', label)
        for k, stu in Students.items():
            r_out, u_out = stu.step(input)
            Rcd_eval.rec_single(k, u_out)
            Rcd_eval.rec_single(k+'r', stu.layers[0].r)

Rcd_eval.finish()



fig, axs = plt.subplots(n_out+2, 1, figsize=(10, 6), sharex=True)
x = np.arange(0, eval_phase, dt)
axs[0].plot(x, Rcd_eval['input'].mean(axis=0), color=black, label='input', linewidth=2, alpha=0.8)

axs[1].plot(x, Rcd_eval['teacher'][0], label="Sum", color=darkred, alpha=0.3, linewidth=4)

    
for i in range(n_out):
    axs[i+1].plot(x, Rcd_eval['naive'][i][:], label="naive", color = lightblue, linewidth=2)
    axs[i+1].plot(x, Rcd_eval['LEEP'][i][:], label="LEEP", color = blue, linewidth=2)
    axs[i+1].plot(x, Rcd_eval['LEEP1'][i][:], label="LEEP1", color = green, linewidth=2)
    axs[i+1].plot(x, Rcd_eval['LEEP2'][i][:], label="LEEP2", color = purple, linewidth=2)
    axs[i+1].plot(x, Rcd_eval['eProp'][i][:], label="eProp", color = pink, linewidth=2)
axs[0].legend()
axs[1].legend()

for i in range(n_rec):
    axs[-1].plot(x, Rcd_eval['naiver'][i][:], label="naive", color = lightblue, linewidth=2)
    #axs[-1].plot(x, Rcd_eval['LEEPr'][i][:], label="LEEP", color = blue, linewidth=2)
    #axs[-1].plot(x, Rcd_eval['LEEP1r'][i][:], label="LEEP1", color = green, linewidth=2)
    #axs[-1].plot(x, Rcd_eval['LEEP2r'][i][:], label="LEEP2", color = purple, linewidth=2)
    #axs[-1].plot(x, Rcd_eval['ePropr'][i][:], label="eProp", color = pink, linewidth=2)

axs[-1].set_xlabel("t (ms)")

axs[0].set_ylabel(r"$r_{in}$")
axs[1].set_ylabel(r"$r_{even}$")

for ax in axs:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
#plt.savefig('./fig/CompareAftRec.png')
plt.show()
```
